```python
import requests, pandas as pd
from bs4 import BeautifulSoup
url="https://en.wikipedia.org/wiki/List_of_Falcon_9_and_Falcon_Heavy_launches"
r=requests.get(url,headers={"User-Agent":"Mozilla/5.0 (educational data analysis)"},timeout=45);r.raise_for_status()
soup=BeautifulSoup(r.text,"html.parser")
print("HTML tables:",len(soup.select("table")))
```

```python
from io import StringIO
tables=pd.read_html(StringIO(r.text))
print("Parsed tables:",len(tables))
for i,t in enumerate(tables[:3]): print("Table",i,t.shape);display(t.head())
# Table layouts may change; choose and normalize the launch table before analysis.
```
