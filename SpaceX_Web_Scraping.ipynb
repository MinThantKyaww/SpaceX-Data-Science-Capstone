# SpaceX Web Scraping

Run all cells to generate outputs. Historical IBM course datasets are included in this repository.

```python
import requests, pandas as pd
from bs4 import BeautifulSoup
url="https://en.wikipedia.org/wiki/List_of_Falcon_9_and_Falcon_Heavy_launches"
r=requests.get(url,headers={"User-Agent":"Mozilla/5.0 (educational data analysis)"},timeout=45);r.raise_for_status()
soup=BeautifulSoup(r.text,"html.parser")
print("HTML tables:",len(soup.select("table")))
```

```python
from io import StringIO
tables=pd.read_html(StringIO(r.text))
print("Parsed tables:",len(tables))
for i,t in enumerate(tables[:3]): print("Table",i,t.shape);display(t.head())
```

```python

import pandas as pd

launch_tables = []

for table in tables:
    if "Flight No." in table.columns and "Launch outcome" in table.columns:
        clean = table.copy()
        clean["Flight No."] = pd.to_numeric(
            clean["Flight No."], errors="coerce"
        )
        clean = clean.dropna(subset=["Flight No."])
        launch_tables.append(clean)

if launch_tables:
    launches_df = pd.concat(launch_tables, ignore_index=True)

    launches_df = launches_df.loc[
        :, ~launches_df.columns.str.startswith("Unnamed:")
    ]

    launches_df = launches_df.drop_duplicates(
        subset=["Flight No."], keep="first"
    )

    print("Scraped launch records:", len(launches_df))
    display(launches_df.head())

    launches_df.to_csv(
        "spacex_web_scraped_launches.csv",
        index=False
    )

    print("Saved spacex_web_scraped_launches.csv")
else:
    print("No matching launch tables found.")
```
