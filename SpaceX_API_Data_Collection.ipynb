# SpaceX API Data Collection


```python
import pandas as pd

url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/dataset_part_1.csv"

launches = pd.read_csv(url)

print("SpaceX Historical Launch Data")
print("----------------------------")
print("Total launch records:", len(launches))
print("Total columns:", len(launches.columns))

print("\nFirst 5 launch records:")
display(launches.head())

print("\nMissing values:")
print(launches.isnull().sum())

filename = "spacex_historical_launches.csv"
launches.to_csv(filename, index=False)

from google.colab import files
files.download(filename)
```
